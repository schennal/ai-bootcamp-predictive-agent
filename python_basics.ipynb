```python
prod1 = {
    "name":"smart watch xyz",
    "price":3000
}

prod2 = {
    "name":"smartphone 11",
    "price":40000
}

def get_discounted_amount(price, disc_perc):
  discounted_price = price - (price * disc_perc/100)
  return discounted_price

disc = 10
prod1_disc_price = get_discounted_amount(prod1["price"], disc)
print(prod1_disc_price)

prod2_disc_price = get_discounted_amount(prod2["price"], disc)
print(prod2_disc_price)
```
